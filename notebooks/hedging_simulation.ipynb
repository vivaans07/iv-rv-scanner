# Delta-hedging P&L on simulated paths

How much does a short straddle earn when realized vol is below implied, and how noisy is it? Uses the same hedger as the paper book.

```python
from volscan.config import PaperSettings
from volscan.paper.simulate import simulate_hedged_straddle

cfg = PaperSettings()
rows = []
for realized in (0.15, 0.20, 0.25, 0.30, 0.35):
    res = simulate_hedged_straddle(
        200, 20, implied_vol=0.25, realized_vol=realized, label="RICH", cfg=cfg
    )
    rows.append(
        (realized, res["gross_pnl"].mean(), res["gross_pnl"].std(), res["hedging_error"].std())
    )
for r in rows:
    print("realized {:.0%}: mean {:+8.0f}  sd {:7.0f}  hedging-error sd {:6.0f}".format(*r))
```
